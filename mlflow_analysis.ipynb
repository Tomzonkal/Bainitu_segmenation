```python
import mlflow
import pandas as pd

# Set the experiment name or ID
experiment_name = "RF Optuna Experiment"
experiment = mlflow.get_experiment_by_name(experiment_name)
experiment_id = experiment.experiment_id

# Search runs (can filter and order_by if needed)
runs = mlflow.search_runs(experiment_ids=[experiment_id], output_format="pandas")

# Only keep the parameter columns
param_columns = [col for col in runs.columns if col.startswith("params.")]
params_df = runs[["run_id"] + param_columns]

# Save to CSV
params_df.to_csv("mlflow_registered_params.csv", index=False)

print("CSV file saved as 'mlflow_registered_params.csv'")
```
